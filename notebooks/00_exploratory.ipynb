# Load the Strathcona data
Reads the parquet files directly

```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

production = pd.read_parquet("../data/daily_production.parquet")
failures = pd.read_parquet("../data/production_failures.parquet")
jobs = pd.read_parquet("../data/well_jobs.parquet")

for name, df in [("production", production), ("failures", failures), ("jobs", jobs)]:
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]} cols")
```

## Missing values
% empty per column

```python
# % of rows where column is blank, ie if theres a 100% on a column than there is no data on that column
for name, df in [("production", production), ("failures", failures), ("jobs", jobs)]:
    print(name)
    print((df.isna().mean() * 100).round(1).sort_values(ascending=False).to_string(), "\n")
```
